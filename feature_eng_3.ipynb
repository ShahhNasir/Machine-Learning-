# KNN Imputer

The KNN Imputer is a method for filling in missing values in a dataset using the K-Nearest Neighbors (KNN) algorithm. It works by finding the k most similar instances (rows) to the one with missing data and imputing the missing values based on the values from those neighbors.

## How it Works?

1. Find k nearest neighbors (based on other features without missing values).
2. Calculate the mean (or weighted average) of the same feature in those neighbors.
3. Impute the missing value with that result.

## When To Use?

1. Data is missing at random.
2. Features are numerical or standardized.
3. Dataset size is manageable (KNN can be slow for large datasets).

```python
# working 
```

dist(x,y) = sqrt(weight * sq. distance from present coordinates)

weight = Total # of coordinates / # of present coordinates

For example, the distance between [3, na, na, 6] and [1, na, 4, 5] is:

 


root(4/2(3-1)**2 + (6-5)**2)

```python
import pandas as pd
```

```python
df = pd.read_csv('train.csv')
```

```python
df.head()
```

```python
new = df[['Age','Pclass','Fare','Survived']]
```

```python
new.head()
```

```python
x = new.drop('Survived',axis=1)
y = new['Survived']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.impute import KNNImputer
```

```python
knn = KNNImputer()

x_train_trf = knn.fit_transform(x_train)
x_test_trf = knn.transform(x_test)
```

```python
pd.DataFrame(x_train_trf,columns=x_train.columns)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr = LogisticRegression()

lr.fit(x_train_trf,y_train)
y_pred = lr.predict(x_test_trf)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y_test,y_pred)
```
