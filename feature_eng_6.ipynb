# Column Transformer

The ColumnTransformer is a powerful tool in sklearn.compose used to apply different preprocessing techniques to different columns of a dataset — all in a single step.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv('covid_toy.csv')
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
x = df.drop('has_covid',axis=1)
```

```python
y = df['has_covid']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.compose import ColumnTransformer
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OrdinalEncoder
from sklearn.preprocessing import OneHotEncoder
```

```python
transformer = ColumnTransformer(transformers=[
    ('tnf1',SimpleImputer(),['fever']),
    ('tnf2',OrdinalEncoder(),['cough']),
    ('tnf3',OneHotEncoder(sparse_output=False,drop='first'),['gender','city'])
    
],remainder='passthrough')
```

```python
transformer.fit_transform(x_train)
```

```python
transformer.transform(x_test)
```

```python
transformer.named_transformers_['tnf3'].get_feature_names_out()
```

```python
ohe = transformer.named_transformers_['tnf3']
ohe.get_feature_names_out(['gender', 'city'])
```

```python
sm = transformer.named_transformers_['tnf1']
sm.get_feature_names_out(['fever'])
```

```python
sm = transformer.named_transformers_['tnf2']
sm.get_feature_names_out(['cough'])
```

# Pipelines

Pipelines chains together mutliple steps so that the ouptput of each step is used as input to the next step.
Pipelines makes it easy to apply the preprocessing to train and test.

```python
new = pd.read_csv('train.csv')
```

```python
new.head()
```

```python
new.drop(columns=['PassengerId','Name','Ticket','Cabin'],axis=1,inplace=True)
```

```python
new.head()
```

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test = train_test_split(new.drop('Survived',axis=1),
                                                 new[['Survived']],test_size=0.2,random_state=42)
```

```python
x_train.head()
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import MinMaxScaler
```

```python

# Categorical pipeline
cat_pipe = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(drop='first', sparse_output=False, handle_unknown='ignore'))
])

# Numerical pipeline
num_pipe = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', MinMaxScaler())
])

# Column transformer
preprocessor = ColumnTransformer([
    ('cat', cat_pipe, ['Sex', 'Embarked']),
    ('num', num_pipe, ['Age', 'Fare'])
], remainder='passthrough')
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
pipe = Pipeline([
    ('preprocess', preprocessor),
    ('model', DecisionTreeClassifier())
])
```

```python
pipe.fit(x_train,y_train)
```

```python
y_pred = pipe.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y_test,y_pred)
```
