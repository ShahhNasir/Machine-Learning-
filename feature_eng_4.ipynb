# One Hot Encoding

One-Hot Encoding is a method used to convert categorical variables into a numerical format that machine learning algorithms can work with.

### Dummy Variable Trap

The dummy variable trap is a scenario in regression modeling (especially linear regression) where multicollinearity arises due to redundant one-hot encoded (dummy) variables.

When you one-hot encode a categorical variable with k categories, you get k binary columns. However, one of those columns is always predictable from the others (i.e., it's linearly dependent). his redundancy causes perfect multicollinearity, which can:
Confuse linear regression models Lead to unstable or non-unique coefficient estimates

```python
df = pd.read_csv('cars.csv')
```

```python
df.head()
```

```python
df = df[['km_driven','fuel','selling_price']]
```

```python
df.head()
```

```python
x_train,x_test,y_train,y_test = train_test_split(df.drop('selling_price',axis=1),
                                                 df['selling_price'],test_size=0.2,random_state=42)
```

```python
df['fuel'].value_counts()
```

# OneHotEncoding using pandas

```python
pd.get_dummies(df,columns=['owner','fuel'],dtype=int)
```

```python
pd.get_dummies(df,columns=['fuel','owner'],drop_first=True,dtype=int)
```
